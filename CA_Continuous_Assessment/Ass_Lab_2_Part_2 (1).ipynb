```python

# Define the map using a graph
map_graph = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D'],
    'D': ['B', 'C']
}

# Define the available colors
colors = ['Red', 'Green', 'Blue']

print("\nMap Representation:")
for region, neighbors in map_graph.items():
    print(region, "->", neighbors)

print("\nAvailable Colors:", colors)
```

```python

def is_safe(region, color, color_assignment, map_graph):
    for neighbor in map_graph[region]:
        if color_assignment.get(neighbor) == color:
            return False
    return True

print("Color validity checking function defined successfully.")
```

```python

# Check whether a color can be assigned to a region
def is_safe(region, color, color_assignment, map_graph):
    for neighbor in map_graph[region]:
        if color_assignment.get(neighbor) == color:
            return False
    return True


# Solve the map coloring problem using backtracking
def map_coloring(map_graph, colors, color_assignment, regions):
    if len(color_assignment) == len(regions):
        return True

    region = regions[len(color_assignment)]

    for color in colors:
        if is_safe(region, color, color_assignment, map_graph):
            color_assignment[region] = color

            if map_coloring(map_graph, colors, color_assignment, regions):
                return True

            # Backtrack if the assignment does not lead to a solution
            del color_assignment[region]

    return False


print("Color validation and backtracking functions defined successfully.")
```

```python

# Initialize an empty color assignment
color_assignment = {}

# Get the list of regions
regions = list(map_graph.keys())

# Execute the map coloring algorithm
if map_coloring(map_graph, colors, color_assignment, regions):
    print("Map Coloring Solution:")
    for region, color in color_assignment.items():
        print(f"Region {region}: {color}")
else:
    print("No valid coloring exists with the given colors.")
```

```python

import networkx as nx
import matplotlib.pyplot as plt

# Create an undirected graph
G = nx.Graph()

# Add regions and their neighboring connections
for region, neighbors in map_graph.items():
    for neighbor in neighbors:
        G.add_edge(region, neighbor)

# Arrange the regions for a neat layout
pos = {
    'A': (0, 1),
    'B': (1.5, 2),
    'C': (1.5, 0),
    'D': (3, 1)
}

# Use the colors assigned by the backtracking algorithm
node_colors = [color_assignment[region].lower() for region in G.nodes()]

plt.figure(figsize=(9, 5.5))

nx.draw_networkx_edges(
    G, pos,
    width=2.5,
    edge_color="gray"
)

nx.draw_networkx_nodes(
    G, pos,
    node_color=node_colors,
    node_size=2400,
    edgecolors="black",
    linewidths=1.5
)

nx.draw_networkx_labels(
    G, pos,
    font_size=15,
    font_weight="bold",
    font_color="white"
)

plt.title(
    "Map Coloring Using Backtracking",
    fontsize=16,
    fontweight="bold",
    pad=20
)

plt.axis("off")
plt.tight_layout()
plt.show()
```
